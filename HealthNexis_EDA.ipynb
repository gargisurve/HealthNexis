**GRAPHICAL UNIVARIATE ANALYSIS**


---



```python
# Count plot for Gender
sns.countplot(x=df['Gender'], palette='gist_rainbow')
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=30)
plt.show()
```

```python
# Count plot for Occupation
sns.countplot(x=df['Occupation'], palette='cool')
plt.title("Occupation Distribution")
plt.xlabel("Occupation")
plt.ylabel("Count")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
junk_counts = df['JunkFood'].value_counts()
print("Junk Food Consumption Counts:\n", junk_counts)
plt.figure(figsize=(7,5))
sns.countplot(x='JunkFood', data=df, palette="hot")
plt.title("Junk Food Consumption Frequency", fontsize=14)
plt.xlabel("Junk Food Consumption", fontsize=12)
plt.ylabel("Number of Participants", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(6,6))
e=[0.05,0.05,0.05]
df['Alcohol'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90,
                                      explode=e, colors=sns.color_palette("gist_rainbow"), shadow=True)
plt.title("Distribution of Alcohol Consumption", fontsize=14)
plt.show()
```

```python
sns.countplot(x='FamilyHistory', data=df, palette="bright")
plt.title("Family History of Diseases", fontsize=14)
plt.xlabel("Family History (Diabetes / BP / Heart Disease)")
plt.ylabel("Count")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**GRAPHICAL BIVARIATE ANALYSIS**

---



```python
plt.figure(figsize=(7,5))
sns.boxplot(x='Occupation', y='StressLevel', data=df, palette="Set3")
plt.title("Stress Levels Across Occupations", fontsize=14)
plt.xlabel("Occupation", fontsize=12)
plt.ylabel("Stress Level (1 = Very Low, 5 = Very High)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6, axis='y')
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.barplot(x='Occupation', y='BMI', data=df, estimator='mean', palette="viridis")
plt.title("Average BMI by Occupation", fontsize=14)
plt.xlabel("Occupation", fontsize=12)
plt.ylabel("Average BMI", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='Gender', y='SleepHours', data=df, palette="pastel")
plt.title("Distribution of Sleep Hours by Gender", fontsize=14)
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Sleep Hours", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.pointplot(x='JunkFood', y='StressLevel', data=df, palette="Dark2", markers="o", linestyles="-")
plt.title("Trend of Stress Level by Junk Food Consumption")
plt.xlabel("Junk Food Consumption")
plt.ylabel("Average Stress Level")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
alcohol_family_pct = pd.crosstab(df['Alcohol'], df['FamilyHistory'], normalize='index') * 100
alcohol_family_pct.plot(kind='bar', figsize=(7,5), colormap="plasma")
plt.xlabel("Alcohol Consumption")
plt.ylabel("Percentage %")
plt.title("Family History among Alcohol Consumers")
plt.legend(title="Family History", labels=["No", "Yes", "Not Sure"], loc='upper center')
plt.xticks(rotation=0)
plt.show()
```

```python
diet_occ = pd.crosstab(df['Occupation'], df['Diet'])

plt.figure(figsize=(7,6))
sns.heatmap(diet_occ, annot=True, cmap="YlGnBu", fmt="d")

plt.title("Heatmap of Daily Diet vs Occupation", fontsize=14)
plt.xlabel("Daily Diet")
plt.ylabel("Occupation")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.violinplot(x='JunkFood', y='Age', data=df, palette="Set2")
plt.title("Age Distribution across Junk Food Consumption Categories", fontsize=14)
plt.xlabel("Junk Food Consumption")
plt.ylabel("Age")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
summary = df.groupby("Alcohol")[["SleepHours","StressLevel"]].mean().reset_index()
summary.plot(x="Alcohol", kind="bar", figsize=(7,6), rot=0, width=0.7, color=["skyblue","salmon"])

plt.title("Alcohol Consumption vs Average Sleep Hours & Stress Level", fontsize=14)
plt.ylabel("Average Value")
plt.xlabel("Alcohol Consumption Category")
plt.legend(["Sleep Hours","Stress Level"])
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
